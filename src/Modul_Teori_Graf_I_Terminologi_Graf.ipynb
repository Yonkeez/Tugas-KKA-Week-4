**NOMOR 2 : [DegreeMaxxing]**
Anda diberikan sebuah daftar ketetanggan sebagai berikut


```python
adj = {
    "A" : {"B", "C","E"},
    "B" : {"A","C","D"},
    "C" : {"A","B","D","E"},
    "D" : {"B","C","E","F"},
    "E" : {"A","C","D","F"},
    "F" : {"D","E"}
}
```

Anda diminta untuk menyelesaikan bagian rumpang dari kode fungsi `count_degree` di mana fungsi ini akan menghitung nilai derajat dari sebuah verteks v berdasarkan daftar ketetanggan yang diberikan

```python
def count_degree(adj, v):
  return len(adj[v])

X = input()
print(count_degree(adj, X))
```

```python
def total_degree(adj):
  return sum(len(neighbors) for neighbors in adj.values())

print (total_degree(adj))
```

```python
def count_edges(adj):
  return total_degree(adj) // 2

print(count_edges(adj))
```

**NOMOR 3 : [NLP Is Very Fun, But too Early for Learn!]** <br>
image.png



Sebuah model bahasa dapat menghasilkan kalimat berdasarkan kata – kata yang ditunjukkan oleh diagram di atas, namun penyusunan kata pada kalimat harus dilakukan secara berurutan berdasarkan tanda panah. Jika ada tanda panah dari kata A ke kata B, artinya di dalam kalimat tersebut, kata B hanya boleh muncul setelah kata A muncul.

Sekarang anda diberikan tantangan untuk menyajikan diagram kata di atas ke dalam daftar ketetanggan di bawah ini :

```python
sentence = "Ubur Ikan Lele Bebek Tegraf Strukdat NLP Is Cool"
sentence = sentence.split()
adj = dict()
# Lengkapi kode rumpang pada fungsi add_connection di bawah ini!
def add_connection(adj: dict, u, v):
  if u not in adj:
    adj[u] = []
  adj[u].append(v)


# Kemudian lanjutkan representasi edge di bawah ini ...
add_connection(adj, sentence[0], sentence[1])  # Ubur -> Ikan
add_connection(adj, sentence[1], sentence[6])
add_connection(adj, sentence[2], sentence[3])
add_connection(adj, sentence[2], sentence[7])
add_connection(adj, sentence[2], sentence[1])
add_connection(adj, sentence[3], sentence[4])
add_connection(adj, sentence[4], sentence[7])
add_connection(adj, sentence[7], sentence[8])
add_connection(adj, sentence[5], sentence[7])
add_connection(adj, sentence[6], sentence[7])
```

Sekarang dilakukan tahap evaluasi / testing pada model bahasa tersebut, anda diberikan input berupa sebuah kata **S** coba anda tentukan kira - kira dengan kata **S**, kalimat terpanjang apa yang bisa dihasilkan?,
** Clue : Degree + Path + Backtracking / Recursive **

```python
# Rekursif
def longest_path(adj, node, visited):
  visited = visited + [node]
  longest = visited

  for neighbors in adj.get(node, []):
    if neighbors not in visited:
      candidate = longest_path(adj, neighbors, visited)
      if len(candidate) > len(longest):
        longest = candidate

  return longest
```

```python
S = input()
result = longest_path(adj, S, [])
print("Kalimat:", " ".join(result))
```

Di lautan berbahaya bernama Grand Line, kompas biasa tidak dapat digunakan karena setiap pulau memiliki medan magnet yang berbeda. Akibatnya, arah menjadi tidak stabil dan tidak bisa ditentukan secara konvensional.

Untuk itu, para navigator menggunakan alat bernama Log Pose. Log Pose akan merekam medan magnet suatu pulau dan menunjuk ke pulau berikutnya, sehingga kapal hanya dapat berpindah dari satu pulau ke pulau lain mengikuti jalur tertentu

Setelah mencapai suatu pulau, Log Pose membutuhkan waktu untuk menyesuaikan diri sebelum dapat menunjuk ke tujuan berikutnya. Hal ini membuat perjalanan di Grand Line bersifat bertahap dan mengikuti jalur yang telah terbentuk


Seorang navigator ingin memetakan jalur pelayaran antar pulau di Grand Line.

Diketahui hubungan antar pulau sebagai berikut:

- Loguetown → Water 7 dengan waktu 5 hari  
- Water 7 → Dressrosa dengan waktu 3 hari  
- Water 7 → Skypiea dengan waktu 2 hari  
- Loguetown → Skypiea dengan waktu 3 hari  
- Loguetown → Wano Country dengan waktu 5 hari  
- Water 7 → Fish-Man Island dengan waktu 2 hari  
- Water 7 → Wano Country dengan waktu 2 hari  
- Loguetown → Egghead dengan waktu 1 hari  



Bantulah seorang navigator untuk menyusun daftar ketetanggaan antar pulau di Grand Line menggunakan program Python.

Setiap hubungan antar dua pulau, yaitu `u` dan `v`, memiliki waktu tempuh `w` yang dapat direpresentasikan sebagai graf berbobot.

Gunakan fungsi `add_connection(u, v, w)` untuk menyimpan hubungan antar pulau ke dalam struktur data adjacency list.

```python
def add_connection(adj: dict, u, v, w):
  if u not in adj:
    adj[u] = []
  if v not in adj:
    adj[v] = []
  adj[u].append((v, w))
  adj[v].append((u, w))

adj = dict()

add_connection(adj, "Loguetown", "Water 7", 5)
add_connection(adj, "Water 7", "Dressrosa", 3)
add_connection(adj, "Water 7", "Skypiea", 2)
add_connection(adj, "Loguetown", "Skypiea", 3)
add_connection(adj, "Loguetown", "Wano Country", 5)
add_connection(adj, "Water 7", "Fish-Man Island", 2)
add_connection(adj, "Water 7", "Wano Country", 2)
add_connection(adj, "Loguetown", "Egghead", 5)

print("Adjacency List Jalur Antar Pulau di Grand Line:")
for island, connections in adj.items():
    print(f"{island} -> {connections}")
```

**NOMOR 6 : [Undirected Graph]** <br>
Sebuah kota direpresentasikan sebagai Graf Tidak Berarah dengan persimpangan sebagai node dan jalan sebagai edge. Graf ini disimpan dalam bentuk Adjacency Matrix.

Diberikan kode berikut, lengkapi bagian yang kosong (...) agar program dapat:

* Membaca jumlah N (jumlah simpul/persimpangan).
* Membaca jumlah M (jumlah jalan).
* Membangun Adjacency Matrix untuk graf berdasarkan pasangan simpul yang terhubung.
* Mencetak Adjacency Matrix yang merepresentasikan jaringan jalan.

```python
N = int(input())
M = int(input())

# Inisialisasi adjacency matrix dengan 0
graph = [[0] * N for _ in range(N)]

for _ in range(M):
    u, v = map(int, input().split())
    graph[u][v] = 1
    graph[v][u] = 1

for row in graph:
    print(" ".join(map(str, row)))
```
